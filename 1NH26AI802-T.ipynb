```python

BITS = 16
MASK = (1 << BITS) - 1


def to_binary(n, bits):
    return format(n & ((1 << bits) - 1), f'0{bits}b')


def signed_value(n):
    if n & (1 << (BITS - 1)):
        return n - (1 << BITS)
    return n


def calculate(a, b, operator):
    print("\nExpression:", f"{a:+d} {operator} ({b:+d})")

   
    a16 = a & MASK
    b16 = b & MASK

   
    if operator == '-':
        operand2 = (~b16 + 1) & MASK
    else:
        operand2 = b16

    result16 = (a16 + operand2) & MASK
    result = signed_value(result16)

    # Display operands in 8-bit and 16-bit formats
    print("Operand A (8-bit) :", to_binary(a, 8))
    print("Operand B (8-bit) :", to_binary(b, 8))
    print("Operand A (16-bit):", to_binary(a16, 16))
    print("Operand B (16-bit):", to_binary(b16, 16))

    if operator == '-':
        print("2's complement of B:", to_binary(operand2, 16))

    print("Result (16-bit)   :", to_binary(result16, 16))
    print("Result (decimal)  :", result)

   
    overflow = result < -32768 or result > 32767
    print("Overflow:", "Yes" if overflow else "No")


print("16-bit 2's Complement Signed Arithmetic")

calculate(56, 29, '+')
calculate(-38, -23, '-')
```
