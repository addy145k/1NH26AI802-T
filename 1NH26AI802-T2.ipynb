```python

def rotate_right(value, count, bits=8):
    mask = (1 << bits) - 1
    value &= mask

    print("\nROR operation")
    print("Original:", format(value, '08b'))

    for i in range(count):
        lsb = value & 1
        value = (value >> 1) | (lsb << (bits - 1))
        print("Step", i + 1, ":", format(value, '08b'))

    return value


def rotate_right_carry(value, count, carry=0, bits=8):
    mask = (1 << bits) - 1
    value &= mask

    print("\nRCR operation")
    print("Original:", format(value, '08b'))
    print("Initial carry:", carry)

    for i in range(count):
        old_lsb = value & 1
        value = (value >> 1) | (carry << (bits - 1))
        carry = old_lsb

        print("Step", i + 1, ":",
              format(value, '08b'),
              "Carry:", carry)

    print("Final result:", format(value, '08b'))
    print("Final carry:", carry)
    return value, carry


ror_result = rotate_right(0b11011101, 3)
print("ROR final result:", format(ror_result, '08b'))

rcr_result, final_carry = rotate_right_carry(
    0b11010001, 3, carry=0
)
```
